## 2.Upload Image for Traning

```python
from google.colab import files
import shutil
import os
from IPython.display import Image, display

# Define the name for the user and the directory
user_name = input("Enter the name for the user: ")
KNOWN_FACES_DIR = "Training_Images"

# Create the user-specific sub-directory if it doesn't exist
user_dir = os.path.join(KNOWN_FACES_DIR, user_name)
if not os.path.exists(user_dir):
    os.makedirs(user_dir)
    print(f"Created directory for {user_name}: {user_dir}")

print(f"Please upload an image for {user_name} (e.g., {user_name}.jpg):")
uploaded = files.upload()

for filename in uploaded.keys():
  # Define the path where the uploaded image will be saved in the training directory
  destination_path = os.path.join(user_dir, filename)

  # Move the uploaded file from the current directory to the destination path
  shutil.move(filename, destination_path)
  print(f'Moved "{filename}" to "{destination_path}"')

  # Display the uploaded image
  display(Image(destination_path))
```

## 3.Encode the test Image

```python
import face_recognition
import os

known_faces = []
known_names = []

# Ensure KNOWN_FACES_DIR is defined
KNOWN_FACES_DIR = "Training_Images"
if not os.path.exists(KNOWN_FACES_DIR):
    print(f"Error: KNOWN_FACES_DIR '{KNOWN_FACES_DIR}' does not exist. Please create it and add face images.")
else:
    for name in os.listdir(KNOWN_FACES_DIR):
        if name.startswith('.'): # Skip hidden files like .DS_Store
            continue
        person_dir = os.path.join(KNOWN_FACES_DIR, name)

        if os.path.isdir(person_dir):
            # This path expects subdirectories for each person
            for filename in os.listdir(person_dir):
                if filename.startswith('.'): continue
                image_path = os.path.join(person_dir, filename)
                print(f"Loading known face image: {image_path}")
                image = face_recognition.load_image_file(image_path)
                encodings = face_recognition.face_encodings(image)
                if encodings:
                    known_faces.append(encodings[0])
                    known_names.append(name) # Use folder name as person's name
                else:
                    print(f"No face found in {image_path}")
        elif os.path.isfile(person_dir):
            # This path is for files directly in KNOWN_FACES_DIR (like previous dummy images or manually uploaded)
            image_path = person_dir
            print(f"Loading known face image: {image_path}")
            image = face_recognition.load_image_file(image_path)
            encodings = face_recognition.face_encodings(image)
            if encodings:
                known_faces.append(encodings[0])
                known_names.append(os.path.splitext(name)[0]) # Use filename (without extension) as person's name
            else:
                print(f"No face found in {image_path}")

print(f"Found {len(known_faces)} known faces.")
print("Known names:", known_names)
```

## 4.Test by Uploading Image

```python
from google.colab import files
import face_recognition
import cv2
import numpy as np
from IPython.display import Image, display

print("Please upload an image for face recognition:")
uploaded_image_files = files.upload()

# Assuming only one image is uploaded for recognition for simplicity
if uploaded_image_files:
    uploaded_filename = list(uploaded_image_files.keys())[0]
    print(f'Uploaded: {uploaded_filename}')

    # Read the image bytes into a numpy array
    image_bytes = uploaded_image_files[uploaded_filename]
    nparr = np.frombuffer(image_bytes, np.uint8)
    unknown_image_bgr = cv2.imdecode(nparr, cv2.IMREAD_COLOR)
    unknown_image_rgb = cv2.cvtColor(unknown_image_bgr, cv2.COLOR_BGR2RGB)

    # Find all the faces in the uploaded image
    face_locations = face_recognition.face_locations(unknown_image_rgb)
    face_encodings = face_recognition.face_encodings(unknown_image_rgb, face_locations)

    display_image = unknown_image_bgr.copy() # Make a copy for drawing

    if face_encodings:
        print(f"Found {len(face_encodings)} face(s) in the uploaded image.")
        for (top, right, bottom, left), face_encoding in zip(face_locations, face_encodings):
            name = "Unknown"

            # Compare current face with known faces
            matches = face_recognition.compare_faces(known_faces, face_encoding)

            if True in matches:
                first_match_index = matches.index(True)
                name = known_names[first_match_index]

            # Draw a box around the face and label it
            cv2.rectangle(display_image, (left, top), (right, bottom), (0, 255, 0), 2)
            cv2.putText(display_image, name, (left + 6, bottom - 6), cv2.FONT_HERSHEY_DUPLEX, 0.7, (255, 255, 255), 1)

        # Convert BGR image back to JPEG bytes for display
        _, buffer = cv2.imencode('.jpg', display_image)
        display(Image(data=buffer.tobytes()))
    else:
        print("No faces found in the uploaded image.")
else:
    print("No image was uploaded.")
```
